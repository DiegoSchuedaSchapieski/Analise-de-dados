```python
import numpy as np
import pandas as pd

# Dados de vendas com inconsistências e valores ausentes (NaN)
dados = {
    'Loja': [
        'Centro',
        'Shopping',
        'Centro',
        'Shopping',
        'Aeroporto',
        'Centro',
        'Aeroporto',
        'Shopping',
    ],
    'Vendedor': [
        'Ana',
        'Carlos',
        'Ana',
        np.nan,
        'Beatriz',
        'Carlos',
        'Beatriz',
        'Ana',
    ],
    'Produto': [
        'Notebook',
        'Smartphone',
        'Tablet',
        'Smartphone',
        'Fone',
        'Notebook',
        'Tablet',
        'Fone',
    ],
    'Preco': [
        3500.0,
        2000.0,
        np.nan,
        2000.0,
        150.0,
        3500.0,
        np.nan,
        150.0,
    ],
    'Quantidade': [2, 1, 3, np.nan, 5, 1, 2, 4],
}

df = pd.DataFrame(dados)

print('--- DADOS BRUTOS ---')
print(df)
```

Identificar valores nulos

```python
df.isna()
```

tratar valores ausentes com a Média

```python
#Primeiro vamos descobrir a média
media_salarial = round(df['Preco'].mean(),2)
print(media_salarial)
```

Agoa vamos substituir pela media, na coluna Preco

```python
df['Preco'] = round(df['Preco'].fillna(df['Preco'].mean()),2)
print(df)
```

Agora na coluna Quantidade, vamos substituir por um os valores null

```python
df['Quantidade'] = df['Quantidade'].fillna(1)
print(df)
```

Criação de uma coluna nova

```python
df['Receita_Total'] = df['Preco'] * df['Quantidade']
#print(df['Receita_Total'])
print(df)
```

Agora vamos agrupar dos dados usando a coluna loja

```python
Lojas = round(df.groupby('Loja').mean('Receita_Total'),2)
#print(Lojas)
Lojas = round(df.groupby('Loja').agg({
    'Receita_Total' : 'sum',
    'Preco' : 'mean'
}),2)

print(Lojas)
```

```python
ranking = df.groupby('Loja')['Receita_Total'].sum().reset_index()

print(ranking)

ranking = ranking.sort_values(by='Receita_Total ',ascending=False).reset_index(drop=True)
print(ranking)
```
