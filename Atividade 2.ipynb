```python
import numpy as np
import pandas as pd

# Dados de vendas do E-commerce
dados = {
    'ID_Pedido': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Loja': ['Curitiba', 'São Paulo', 'Curitiba', 'Rio', 'São Paulo', 'Rio', 'Curitiba', 'São Paulo', 'Rio', 'Curitiba'],
    'Categoria': ['Eletrônicos', 'Móveis', 'Eletrônicos', 'Móveis', 'Eletrônicos', 'Eletrônicos', 'Móveis', 'Móveis', 'Eletrônicos', 'Eletrônicos'],
    'Preco_Unitario': [2500.0, 1200.0, np.nan, 800.0, 2500.0, 150.0, np.nan, 1200.0, 150.0, 3000.0],
    'Quantidade': [1, 2, 3, np.nan, 2, 5, 1, np.nan, 4, 1],
    'Desconto_Perc': [0.10, 0.0, np.nan, 0.05, 0.10, 0.0, 0.15, 0.0, np.nan, 0.20]
}

df_vendas = pd.DataFrame(dados)

print("--- DADOS BRUTOS DO E-COMMERCE ---")
print(df_vendas)
```

```python
Media_Vendas = round(df_vendas['Preco_Unitario'].mean(),2)
print(Media_Vendas)
```

```python
df_vendas['Preco_Unitario'] = round(df_vendas['Preco_Unitario'].fillna(df_vendas['Preco_Unitario'].mean()),2)
print(df_vendas)
```

```python
df_vendas['Quantidade'] = df_vendas['Quantidade'].fillna(1)
print(df_vendas)
```

```python
df_vendas['Desconto_Perc'] = df_vendas['Desconto_Perc'].fillna(0)
print(df_vendas)
```

```python
df_vendas['Valor_Bruto'] = df_vendas['Quantidade'] * df_vendas['Preco_Unitario']
print(df_vendas)
```

```python
df_vendas['Valor_Liquido'] = df_vendas['Valor_Bruto'] * (1 - df_vendas['Desconto_Perc'])
print(df_vendas)
```

```python
Lojas = round(df_vendas.groupby('Loja').mean('Valor_Liquido'),2)
Lojas = round(df_vendas.groupby('Loja').agg({
    'Valor_Liquido' : 'sum',
    'Quantidade' : 'mean'
}))

print(Lojas)
```

```python
ranking = df_vendas.groupby('Loja').sum().reset_index()
print(ranking)
```
